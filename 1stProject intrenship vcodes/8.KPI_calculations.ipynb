```python

# Day 29 - KPI Development

import pandas as pd
import numpy as np

# Load transformed dataset using the full file path
df = pd.read_csv(
    "transformed_retail_sales.csv"
)

print("Dataset loaded successfully!")

print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

df.head()
```

```python
#KPI 1 — Total Revenue 
total_revenue=df["Sales"].sum()
print("KPI Name: Total Revenue")
print("Total Revenue:",round(total_revenue,2))
```

```python
# KPI 2: Total Profit
total_profit=df["Profit"].sum()
print("KPI Name: Total Profit")
print("Total Profit:", round(total_profit, 2))
```

```python
# KPI 3 — Total Orders 
total_orders=df['Order ID'].nunique()
print("KPI Name: Total Orders")
print("Total Orders:", total_orders)
```

```python
# KPI 4 — Total Customers 
total_customers=df['Customer ID'].nunique()
print("KPI Name: Total Customers")
print("Total Customers:", total_customers)
```

```python
# KPI 5 — Average Order Value 
average_order_value=total_revenue/ total_orders
print("KPI Name: Average Order Value")
print("Average Order Value:", round(average_order_value, 2))
```

```python
# KPI 6 — Profit Margin
profit_margin=(total_profit/total_revenue)*100
print("KPI Name: Profit Margin")
print("Profit Margin:", round(profit_margin, 2), "%")
```

```python
# KPI 7 — Growth %
# Check available years in the dataset
print("Available Years:")
print(sorted(df["Year"].unique()))
# KPI 7: Revenue Growth
yearly_revenue=df.groupby('Year')['Sales'].sum()
previous_revenue=yearly_revenue.loc[2016]
current_revenue=yearly_revenue.loc[2017]
growth_percentage=(
    (current_revenue-previous_revenue)/previous_revenue
)*100
print("KPI Name: Revenue Growth")
print("Previous Year Revenue (2016):", round(previous_revenue, 2))
print("Current Year Revenue (2017):", round(current_revenue, 2))
print("Revenue Growth:", round(growth_percentage, 2), "%")
```

```python
# Final KPI Summary
kpi_summary = {
    "KPI Name": [
        "Total Revenue","Total Profit",
        "Total Orders","Total Customers",
        "Average Order Value", "Profit Margin","Revenue Growth"],
    "Value": [
        round(total_revenue, 2), round(total_profit, 2), total_orders, total_customers,
        round(average_order_value, 2), round(profit_margin, 2),  round(growth_percentage, 2) ]
}
kpi_df = pd.DataFrame(kpi_summary)
print("Retail Sales Performance - KPI Summary")
display(kpi_df)
```
